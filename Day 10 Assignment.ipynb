8. Compare Photon, Predictive I/O, and Intelligent Workload Management's roles in why a serverless
warehouse can answer ad hoc dashboard queries fast, and use that to justify serverless over
classic/pro warehouses for this use case.

Photon, Predictive I/O, and Intelligent Workload Management

A serverless SQL warehouse can answer ad hoc dashboard queries quickly because several optimizations work together:

* **Photon:** Accelerates SQL query execution by using a highly optimized query engine, improving joins, aggregations, and scans.
* **Predictive I/O:** Predicts which data will be needed and optimizes data access, reducing unnecessary I/O and improving query performance.
* **Intelligent Workload Management (IWM):** Automatically manages compute resources and prioritizes workloads so that concurrent dashboard queries can receive resources efficiently.

Why Serverless?

For an executive dashboard with unpredictable, ad hoc queries, **serverless SQL warehouses** are preferable because they provide fast startup, automatic resource management, and optimized query execution without requiring manual cluster management.

Compared with classic/pro warehouses, serverless reduces operational overhead and can automatically adapt resources to changing query workloads, making it a better fit for interactive dashboard and Genie workloads.

**Conclusion:** Photon provides fast execution, Predictive I/O improves data access, and IWM manages workload resources. Together, they make serverless warehouses well suited for fast, unpredictable executive-facing queries.
